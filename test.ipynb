# Imports

```python
import pandas as pd
import os
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv("data/openagenda_idf_events.csv")
```

# Data exploration

```python
df.info()
```

```python
df = df[df["longdescription_fr"].notna()]
```

```python


lengths = df["longdescription_fr"].dropna().str.len()

plt.figure(figsize=(10, 5))
plt.hist(lengths, bins=50, edgecolor="black")
plt.xlabel("Length of longdescription_fr (characters)")
plt.ylabel("Number of events")
plt.title("Distribution of longdescription_fr length")
plt.show()

lengths.describe()
```

```python
from langchain_mistralai import MistralAIEmbeddings

embeddings = MistralAIEmbeddings(
    model="mistral-embed",
    api_key=os.environ["MISTRAL_API_KEY"],  # get free at console.mistral.ai
)
vectors = embeddings.embed_documents(df["longdescription_fr"].tolist())
```
